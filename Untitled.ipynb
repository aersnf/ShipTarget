```python
%matplotlib inline
import pandas as pd
import matplotlib.pyplot as plt
import os
from pathlib import Path
# 创建项目目录结构
# 获取当前脚本所在目录的绝对路径
current_dir = Path.cwd()  # 或者 Path(".").absolute()
project_root = current_dir / "ship_detection"
dataset_dir = project_root / "datasets" / "SSDD"
models_dir = project_root / "models"
results_dir = project_root / "results"

print(f"项目根目录: {project_root}")
```

```python
# ==================== 配置区域 ====================
# 请修改为您的实际CSV文件路径
csv_files = {
    "YOLOv11": "ship_detection/results/yolov11/baseline_SSDD/results.csv",
    "YOLOv11+MMSA (MuSGD)": "ship_detection/results/improved/mmsa_SSDD/results.csv",
    "YOLOv11+MMSA (AdamW)": "ship_detection/results/improved/mmsa_SSDD_2/results.csv",
    # 如果有YOLOv8/YOLOv10的CSV，也可以添加
}

# 输出图片保存路径
output_loss_curve = "loss_curves.png"
output_map_curve = "map_curves.png"

# 图表样式设置
plt.style.use('seaborn-v0_8-darkgrid')
colors = ['#1f77b4', '#ff7f0e', '#2ca02c', '#d62728', '#9467bd']
line_styles = ['-', '--', '-.', ':']

# ==================== 读取数据 ====================
def load_csv(file_path):
    if not os.path.exists(file_path):
        print(f"警告: 文件不存在 {file_path}")
        return None
    df = pd.read_csv(file_path)
    # 去除空列（如有）
    df = df.loc[:, ~df.columns.str.contains('^Unnamed')]
    return df

# ==================== 绘制损失曲线 ====================
plt.figure(figsize=(12, 5))

# 子图1: 训练损失（box loss, cls loss, dfl loss）
plt.subplot(1, 2, 1)
for idx, (name, path) in enumerate(csv_files.items()):
    df = load_csv(path)
    if df is None:
        continue
    # 检查是否存在train/box_loss列（Ultralytics 常见列名）
    if 'train/box_loss' in df.columns:
        plt.plot(df['epoch'], df['train/box_loss'], 
                 color=colors[idx % len(colors)], 
                 linestyle=line_styles[idx % len(line_styles)],
                 label=f"{name} - box loss")
    # 可选的 cls_loss 和 dfl_loss
    if 'train/cls_loss' in df.columns:
        plt.plot(df['epoch'], df['train/cls_loss'], 
                 color=colors[idx % len(colors)], 
                 linestyle='--', alpha=0.6,
                 label=f"{name} - cls loss")
plt.xlabel('Epoch')
plt.ylabel('Loss')
plt.title('Training Loss Curves')
plt.legend(loc='upper right', fontsize=8)
plt.grid(True)

# 子图2: 验证损失（如果有）
plt.subplot(1, 2, 2)
for idx, (name, path) in enumerate(csv_files.items()):
    df = load_csv(path)
    if df is None:
        continue
    if 'val/box_loss' in df.columns:
        plt.plot(df['epoch'], df['val/box_loss'], 
                 color=colors[idx % len(colors)], 
                 linestyle=line_styles[idx % len(line_styles)],
                 label=f"{name} - val box loss")
plt.xlabel('Epoch')
plt.ylabel('Loss')
plt.title('Validation Loss Curves')
plt.legend(loc='upper right', fontsize=8)
plt.grid(True)

plt.tight_layout()
plt.savefig(output_loss_curve, dpi=300)
print(f"损失曲线图已保存至: {output_loss_curve}")

# ==================== 绘制mAP曲线 ====================
plt.figure(figsize=(12, 5))

# 子图1: mAP@0.5
plt.subplot(1, 2, 1)
for idx, (name, path) in enumerate(csv_files.items()):
    df = load_csv(path)
    if df is None:
        continue
    if 'metrics/mAP50(B)' in df.columns:
        plt.plot(df['epoch'], df['metrics/mAP50(B)'], 
                 color=colors[idx % len(colors)], 
                 linestyle=line_styles[idx % len(line_styles)],
                 label=name)
plt.xlabel('Epoch')
plt.ylabel('mAP@0.5')
plt.title('mAP@0.5 Comparison')
plt.legend(loc='lower right')
plt.grid(True)

# 子图2: mAP@0.5:0.95
plt.subplot(1, 2, 2)
for idx, (name, path) in enumerate(csv_files.items()):
    df = load_csv(path)
    if df is None:
        continue
    if 'metrics/mAP50-95(B)' in df.columns:
        plt.plot(df['epoch'], df['metrics/mAP50-95(B)'], 
                 color=colors[idx % len(colors)], 
                 linestyle=line_styles[idx % len(line_styles)],
                 label=name)
plt.xlabel('Epoch')
plt.ylabel('mAP@0.5:0.95')
plt.title('mAP@0.5:0.95 Comparison')
plt.legend(loc='lower right')
plt.grid(True)

plt.tight_layout()
plt.savefig(output_map_curve, dpi=300)
print(f"mAP曲线图已保存至: {output_map_curve}")

# ==================== 可选：绘制精确率/召回率曲线 ====================
# 如果需要，可以单独绘制 Precision/Recall 曲线
plt.figure(figsize=(10, 5))
for idx, (name, path) in enumerate(csv_files.items()):
    df = load_csv(path)
    if df is None:
        continue
    if 'metrics/precision(B)' in df.columns:
        plt.plot(df['epoch'], df['metrics/precision(B)'], 
                 color=colors[idx % len(colors)], 
                 linestyle=line_styles[idx % len(line_styles)],
                 label=f"{name} - Precision")
    if 'metrics/recall(B)' in df.columns:
        plt.plot(df['epoch'], df['metrics/recall(B)'], 
                 color=colors[idx % len(colors)], 
                 linestyle='--', alpha=0.6,
                 label=f"{name} - Recall")
plt.xlabel('Epoch')
plt.ylabel('Metric')
plt.title('Precision and Recall Curves')
plt.legend(loc='best', fontsize=8)
plt.grid(True)
plt.tight_layout()
plt.savefig("precision_recall_curves.png", dpi=300)
print("精确率/召回率曲线图已保存至: precision_recall_curves.png")
```
